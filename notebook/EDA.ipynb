```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
    
def preprocess_data(df):
    """
    the perprocessing steps for the data
    Parameters:
        df (pd.DataFrame): the data frame that we want to preprocess
    Returns:
        pd.DataFrame: the preprocessed data frame
    """
    # split the columns
    df[['Group', 'Group_Num']] = df['PassengerId'].str.split('_', expand=True)
    # convet to int
    df['Group'] = df['Group'].astype(int)
    df['Group_Num'] = df['Group_Num'].astype(int)  
    
    # Create Group_size feature
    df['Group_size'] = df.groupby('Group')['Group'].transform('count')
 
    
    # create new columns for is_alone
    df['is_alone'] = df['Group']==1
    
    # drop the Name and PassengerId column
    df.drop(columns=['PassengerId','Name'], inplace=True)
    
    # fill null by ('unknown/0/U') and split the cabin column into three columns (Deck,Num,Side) and drop the cabin column and convert the num column to numeric and create new column for num_bin by using qcut
    df['Cabin'] = df['Cabin'].fillna('Unknown/0/U')
    # split the data 
    df[['Deck', 'Num', 'Side']] = df['Cabin'].str.split('/', expand=True)
    # drop the cabin column
    df.drop(columns=['Cabin'], inplace=True)
    # convert the num column to numeric
    df['Num'] = pd.to_numeric(df['Num'], errors='coerce')
    # create new column for num_bin by using qcut and make in 4 group 
    df['Num_bin'] = pd.qcut(df['Num'], q=4, duplicates='drop')
        
    
    # cabin column we can fill the missing value by ('unknown/0/U') and split the cabin column into three columns (Deck,Num,Side) and drop the cabin column and convert the num column to numeric and create new column for num_bin by using qcut
    df['Cabin'] = df['Cabin'].fillna('Unknown/0/U')
    df['Destination'] = df['Destination'].fillna('Unknown')
    
    # +++++++++++++++++++++++++++++++++++++++++++++++++++++++++
    cat_cols = ['HomePlanet', 'Destination', 'Deck', 'Side', 'Num_bin']
    
    preprocessor = ColumnTransformer(
        transformers=[
            ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
        ],
        remainder='passthrough'
    )
    print(preprocessor.fit_transform(df).shape)
    
    # +++++++++++++++++++++++++++++++++++++++++++++++++++++++++
    
    
    # fill the missing in (VIP , TRansported , CryoSleep) by False
    df[['CryoSleep','VIP','Transported']] = df[['CryoSleep','VIP','Transported']].fillna(False)
     
    #  fill the missing value in column Age by meadin()
    df['Age'] = df['Age'].fillna(df['Age'].median())
   
   
    # fill the missing value in column spending by fillna(0) 
    cols = ['RoomService', 'FoodCourt', 'ShoppingMall', 'Spa', 'VRDeck']
    df[cols] = df[cols].fillna(0)
    # make anther column for total spending by sum the spending columns
    df['Total_spending'] = df[cols].sum(axis=1)
    return df
     
```
